## 3.1 input()-funksjonen

Så langt har vi selv skrevet inn verdiene i koden. Men ofte ønsker vi at *brukeren* av programmet skal kunne skrive inn informasjon selv, mens programmet kjører. Da bruker vi `input()`.

```python
navn = input("Hva heter du? ")
```

Når denne linjen kjører, vil Python skrive ut `Hva heter du?` og vente på at brukeren skriver noe og trykker Enter. Det brukeren skriver blir lagret i variabelen `navn`.

**Viktig:** `input()` gir alltid en **tekststreng** tilbake, selv om brukeren skriver inn et tall!

````python
navn = input("Hva heter du? ")
print("Hei,", navn, "! Hyggelig å møte deg.")
````

### 3.1.1 Endring av datatype fra input

Fordi `input()` alltid gir tekst, får vi problemer hvis vi prøver å regne med det brukeren har skrevet inn direkte.

````python
alder = input("Hvor gammel er du? ")
print(type(alder))          # Denne er 'str', selv om brukeren skrev et tall!

# Dette vil gi en feilmelding fordi vi prøver å legge sammen tekst og tall:
# neste_ar = alder + 1
````

For å kunne regne med verdien, må vi gjøre den om til et tall med `int()` eller `float()`.

````python
alder = input("Hvor gammel er du? ")
alder = int(alder)          # Gjør om til heltall
neste_ar = alder + 1
print("Neste år blir du", neste_ar, "år.")
````

## 3.2 Sammenlikningsoperatorer

Når vi programmerer ønsker vi ofte å sammenlikne to verdier. Resultatet av en sammenlikning er alltid en **boolsk verdi**: `True` eller `False`.

| Operator | Betydning |
|---|---|
| `==` | er lik |
| `!=` | er ikke lik |
| `<` | mindre enn |
| `>` | større enn |
| `<=` | mindre enn eller lik |
| `>=` | større enn eller lik |

**NB:** `=` brukes til å *tilordne* en verdi til en variabel, mens `==` brukes til å *sammenlikne*. Dette er en av de vanligste nybegynnerfeilene!

````python
print(5 == 5)     # True
print(5 == 4)     # False
print(5 != 4)     # True
print(3 < 8)      # True
print(3 > 8)      # False
print(3 <= 3)     # True
````

## 3.3 if, elif og else

Med `if` kan vi få programmet til å utføre kode **bare hvis** en betingelse er sann. Med `else` bestemmer vi hva som skal skje **hvis betingelsen ikke er sann**. Med `elif` (else if) kan vi sjekke flere betingelser etter hverandre.

```python
if betingelse:
    # kode som kjøres hvis betingelsen er True
elif annen_betingelse:
    # kode som kjøres hvis betingelse 1 er False, men denne er True
else:
    # kode som kjøres hvis ingen av betingelsene er True
```

**Viktig:** Legg merke til kolon (`:`) etter betingelsen, og at koden under er rykket inn (indentert). Python bruker innrykk for å vite hva som hører til `if`-blokken.

````python
alder = 17

if alder >= 18:
    print("Du er myndig.")
else:
    print("Du er ikke myndig ennå.")
````

````python
karakter = 4

if karakter >= 5:
    print("Karakteren er høy")
elif karakter >= 3:
    print("Karakteren er middels")
else:
    print("Karakteren er lav")
````

## 3.4 Logiske operatorer: and, or og not

Noen ganger må vi sjekke **flere betingelser samtidig**. Da bruker vi de logiske operatorene:

| Operator | Betydning |
|---|---|
| `and` | Begge betingelsene må være sanne |
| `or` | Minst én av betingelsene må være sann |
| `not` | Snur om på en boolsk verdi (`True` blir `False` og omvendt) |

````python
alder = 20
har_lappen = True

if alder >= 18 and har_lappen:
    print("Du kan kjøre bil.")
else:
    print("Du kan ikke kjøre bil.")
````

````python
dag = "lørdag"

if dag == "lørdag" or dag == "søndag":
    print("Det er helg!")
else:
    print("Det er hverdag.")
````

````python
er_pausert = False
print(not er_pausert)   # True, fordi vi snur om på verdien
````

**Oppgave C**

**Oppgave D**  
Lag et program som spør brukeren om totalsummen på en handlekurv i en nettbutikk. Butikken gir gratis frakt hvis summen er over 500 kr **og** kunden er medlem av kundeklubben. Spør brukeren om begge deler, og skriv ut om de får gratis frakt.

````python
# Oppgave C

num1 = int(input("Skriv tall 1: "))
num2 = int(input("Skriv tall 2: "))

if num1 > num2:
    print("Tall 1 er større enn Tall 2")
elif num1 < num2:
    print("Tall 2 er større enn tall 1")
elif num1 == num2:
    print("Tallene er lik!")
````

````python
# Oppgave D

er_medlem = False
while True:
    x = input("Er du medlem? (ja/nei): ")
    # Bruk .lower() for å håndtere store bokstaver
    if str(x).lower() == "ja" or str(x).lower() == "nei":
        # Condition er True om brukeren svarte ja.
        # Som da betyr at den blir False om brukeren svarte Nei.
        er_medlem = (str(x).lower() == "ja")
        # Pell dæ ut av loopen!
        break

pris = float(input("Hva er prisen på kjøpet ditt?: "))

if pris >= 500 and er_medlem:
    print("Du får gratis frakt!")
else:
    print("Du må betale frakt :(")
````
