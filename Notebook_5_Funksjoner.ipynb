## 5.1 Hva er en funksjon?

En funksjon er en **avgrenset kodebit** som utfører en bestemt oppgave, og som vi kan gjenbruke om og om igjen uten å skrive den samme koden på nytt hver gang.

Du har allerede brukt funksjoner: `print()`, `input()`, `int()`, `type()` er alle eksempler på **innebygde** funksjoner i Python. Nå skal vi lære å lage våre **egne**.

### Oppskrift for å lage en funksjon i Python

```python
def funksjonsnavn(argument1, argument2, ...):
    # koden funksjonen skal gjøre
    return resultat
```

- `def` (definer) forteller Python at nå kommer det en ny funksjon
- Navnet på funksjonen følger rett etter, med parenteser
- Inni parentesene kan vi ha ingen, ett eller flere **argumenter** (verdier funksjonen tar imot)
- `return` bestemmer hva funksjonen skal sende tilbake til den som kalte på den
- Husk kolon (`:`) og innrykk, akkurat som med `if` og løkker

## 5.3 Funksjon med ett argument

## 5.5 Lokale og globale variabler

Når vi lager en variabel **inni** en funksjon, kalles den en **lokal variabel**. Den eksisterer bare inni funksjonen, og "forsvinner" når funksjonen er ferdig å kjøre. Den kan ikke brukes utenfor funksjonen.

En variabel som opprettes **utenfor** alle funksjoner, kalles en **global variabel**. Den kan leses fra hvor som helst i programmet, også inni funksjoner.

````python
def test_funksjon():
    lokal_variabel = "Jeg finnes bare inni funksjonen"
    print(lokal_variabel)

test_funksjon()

# Dette vil gi feilmelding, fordi lokal_variabel ikke finnes utenfor funksjonen:
# print(lokal_variabel)
````

````python
global_variabel = "Jeg finnes i hele programmet"

def annen_funksjon():
    print(global_variabel)   # Vi KAN lese globale variabler inni en funksjon

annen_funksjon()
print(global_variabel)
````

**Hvorfor er dette lurt å vite?** Hvis to funksjoner bruker samme variabelnavn lokalt, påvirker de ikke hverandre. Det gjør koden mer forutsigbar og enklere å feilsøke, fordi en funksjon ikke ved et uhell kan ødelegge data andre steder i programmet.

## 5.6 Funksjoner med ukjent antall argumenter (*args)

Noen ganger vet vi ikke på forhånd hvor mange argumenter en funksjon skal ta imot. Da kan vi bruke `*args` (kort for "arguments"). Da samler Python alle argumentene som blir sendt inn i en liste, som vi kan gå gjennom med en `for`-løkke.

````python
def summer_alle(*args):
    total = 0
    for tall in args:
        total = total + tall
    return total

print(summer_alle(1, 4, 7))
print(summer_alle(1, 4, 7, 6, 3, 1, 5, 6))    # Fungerer med hvor mange tall som helst
````

### 🔧 Oppgave 5.6.1
Lag en funksjon `finn_storste(*args)` som returnerer det største tallet blant argumentene som sendes inn. Test funksjonen med minst 4 tall.

*Tips: du kan bruke den innebygde funksjonen `max()` inni funksjonen din, eller løse det selv med en løkke og en `if`-setning.*

````python
# Skriv koden din her

def biggest(*args):
    nums = []
    for arg in args:
        nums.append(arg)
    biggest = 0
    for num in nums:
        if num > biggest:
            biggest = num
    return biggest

print(biggest(1,2,3,4,5))
print(biggest(1,2,40,4,5))
print(biggest(144,2,0,4,5))
````

## 5.7 lambda-funksjoner

En `lambda`-funksjon er en kort, "anonym" funksjon skrevet på én linje. Den brukes når funksjonen er enkel, og vi ikke trenger å gi den et fullt `def`-oppsett.

```python
funksjonsnavn = lambda argumenter: uttrykk
```

````python
dobbel = lambda x: x * 2
print(dobbel(5))

# Sammenlign med den vanlige måten:
def dobbel_vanlig(x):
    return x * 2

print(dobbel_vanlig(5))
````

````python
summer = lambda a, b: a + b
print(summer(3, 8))
````

## 5.8 Funksjoner og løkker sammen

Funksjoner kan kalles flere ganger, for eksempel inni en `for`-løkke.

````python
def er_partall(tall):
    return tall % 2 == 0

for i in range(1, 11):
    if er_partall(i):
        print(i, "er et partall")
    else:
        print(i, "er et oddetall")
````

### 💡 Eksempel: Billettpris basert på alder
Et vanlig eksempel fra "den virkelige verden" er å regne ut riktig billettype basert på alder. Her kombinerer vi en funksjon med flere `elif`-betingelser.

````python
def billettype(alder):
    alder = int(alder)

    if 6 <= alder <= 17:
        return "Barn/Ungdom"
    elif 18 <= alder <= 29:
        return "Ung voksen"
    elif alder >= 67:
        return "Honnør"
    elif alder < 6:
        return "Gratis"
    else:
        return "Voksen"

print(billettype(46))
print(billettype(9))
print(billettype(70))
````
