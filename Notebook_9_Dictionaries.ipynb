## 9.1 Hva er en dictionary?

En **dictionary** (ordbok) er en datatype som lagrer data i **par**: en **nøkkel** (key) og en **verdi** (value). I stedet for å hente ut data med et plassnummer (som i lister), henter vi ut data ved å bruke nøkkelen.

En dictionary...
- er **ordnet** (rekkefølgen på elementene er bestemt av når de ble lagt til)
- kan **endres** (vi kan legge til, endre og fjerne elementer)
- godtar **ikke duplikate nøkler** (hver nøkkel kan bare finnes én gang)

En dictionary skrives med krøllparenteser `{ }`, og hvert par skrives som `nøkkel : verdi`.

```python
bil = {"merke": "Volkswagen", "modell": "Polo Cross", "år": 2010}
```

Tenk på en dictionary som en ekte ordbok: du slår opp et **ord** (nøkkel) for å finne **forklaringen** (verdi) - ikke et sidetall.

````python
bil = {"merke": "Volkswagen", "modell": "Polo Cross", "år": 2010,
       "farge": "blå", "type": "diesel"}

print(bil)
print(type(bil))
````

## 9.2 Hente ut verdier

Vi henter ut en verdi ved å skrive nøkkelen i hakeparenteser, `dictionary[nøkkel]`. Vi kan også bruke metoden `.get(nøkkel)`, som gjør det samme.

````python
print(bil["merke"])
print(bil["år"])

print(bil.get("merke"))    # Gjør det samme som bil["merke"]
````

### Hente ut alle nøkler eller verdier

- `.keys()` gir alle nøklene i dictionary
- `.values()` gir alle verdiene i dictionary
- `.items()` gir alle nøkkel-verdi-parene

````python
print(bil.keys())
print(bil.values())
print(bil.items())
````

## 9.3 Legge til, endre og fjerne elementer

- **Legge til / endre**: `dictionary[nøkkel] = verdi` - hvis nøkkelen ikke finnes fra før, lages et nytt element. Finnes den fra før, blir verdien overskrevet.
- **Fjerne**: `.pop(nøkkel)` fjerner nøkkelen og verdien

````python
bil["km.stand"] = 65000    # Legger til et nytt element
print(bil)

bil["år"] = 2007            # Endrer verdien til en eksisterende nøkkel
print(bil)

bil.pop("type")             # Fjerner nøkkelen "type" og verdien dens
print(bil)
````

## 9.4 Sjekke om en nøkkel finnes

Vi bruker `in` for å sjekke om en nøkkel finnes i en dictionary, akkurat som med lister.

````python
navn = {"Markus": "Hansen", "Mohammed": "Ali", "Steve": "Mathisen",
        "Emil": "Hansen", "Nikolai": "Pettersen"}

if "Steve" in navn:
    print("Ja, Steve finnes i dictionary navn")
else:
    print("Nei, Steve finnes ikke")
````

## 9.5 Gå gjennom en dictionary med for-løkke

Når vi bruker en `for`-løkke direkte på en dictionary, går vi automatisk gjennom **nøklene**. Vi kan bruke nøkkelen til å hente ut verdien inni løkken.

````python
for key in navn:
    print(key)              # Skriver ut hver nøkkel (fornavn)
````

````python
for key in navn:
    print(key, "->", navn[key])    # Skriver ut nøkkel og tilhørende verdi
````

Vi kan også bruke `.items()` for å hente ut nøkkel og verdi samtidig - dette er ofte litt ryddigere kode.

````python
for key, verdi in navn.items():
    print(key, "har etternavnet", verdi)
````

## 9.6 Nested dictionaries (dictionary i dictionary)

En dictionary kan inneholde andre dictionaries som verdier. Dette kalles en **nested** (nøstet) dictionary, og er nyttig når data har flere nivåer, for eksempel en familie med flere barn, hvor hvert barn har egen informasjon.

````python
barn_1 = {"navn": "Rudolf", "alder": 10}
barn_2 = {"navn": "Nils", "alder": 5}
barn_3 = {"navn": "Kristina", "alder": 7}

familie = {"barn_1": barn_1, "barn_2": barn_2, "barn_3": barn_3}

print(familie["barn_2"])              # Henter ut hele dictionary til barn_2
print(familie["barn_2"]["navn"])      # Henter ut kun navnet til barn_2
````

### 🔴 Vanskelig nivå (utfordrende)

**Oppgave C**  
Lag et Python-program som konverterer to lister til én dictionary, for eksempel:
```python
navn = ["Ola", "Kari", "Nils"]
poeng = [45, 78, 62]
```
skal bli `{"Ola": 45, "Kari": 78, "Nils": 62}`.  
*Tips: søk opp den innebygde funksjonen `zip()`, eller løs det med en `for`-løkke og `range(len(...))`.*

**Oppgave D**  
Lag et program som summerer alle verdiene i en dictionary, for eksempel:
```python
poengsum = {"Ola": 45, "Kari": 78, "Nils": 62}
```
Programmet skal regne ut og skrive ut totalsummen av alle verdiene.

````python
# Oppgave C

navn = ["Ola", "Kari", "Nils"]
poeng = [45, 78, 62]

data = {}

for i in range(len(navn)):
    data[navn[i]] = poeng[i]
print(data)

# Oppgave D

sum = 0
for x in data:
    sum += data[x]
print(sum)
````

## Oppsummering
- En dictionary lagrer data som nøkkel-verdi-par: `{"nøkkel": verdi}`
- Vi henter verdier med `dictionary[nøkkel]` eller `.get(nøkkel)`
- `.keys()`, `.values()` og `.items()` gir oss nøklene, verdiene eller begge deler
- Vi legger til/endrer med `dictionary[nøkkel] = verdi`, og fjerner med `.pop(nøkkel)`
- `in` sjekker om en nøkkel finnes
- En `for`-løkke over en dictionary går gjennom nøklene
- Dictionaries kan nøstes - en verdi kan selv være en dictionary
