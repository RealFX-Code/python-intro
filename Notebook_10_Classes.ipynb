## 10.1 Hva er objektorientert programmering?

Python er et **objektorientert programmeringsspråk**. Det betyr at vi kan organisere koden vår i **klasser** og **objekter**, i stedet for bare variabler og funksjoner etter hverandre.

- En **klasse** er en "oppskrift" eller "mal" som beskriver hvilke egenskaper (data) og handlinger (funksjoner) noe skal ha.
- Et **objekt** er en konkret "instans" laget etter denne oppskriften.

**Eksempel:** Tenk deg at du lager et program for et bibliotek. Da vil det være naturlig å lage en klasse `Bok`, som beskriver at hver bok har en tittel, en forfatter og et utgivelsesår. Hver gang biblioteket registrerer en ny bok, lager vi et nytt **objekt** av klassen `Bok` - med sine egne, unike verdier.

Data og funksjoner (kalt **metoder** når de hører til en klasse) som hører sammen, samles i én enhet: objektet.

### DRY-prinsippet
Et viktig prinsipp i programmering er **DRY** - "Don't Repeat Yourself". Klasser er nyttige nettopp fordi de lar oss samle mye funksjonalitet på ett sted, i stedet for å skrive den samme koden om og om igjen for hver "ting" i programmet vårt.

## 10.2 `__init__()` - klassens "oppstartsfunksjon"

I en klasse har vi en spesiell, innebygd funksjon som heter `__init__()`. Denne blir automatisk kalt hver gang vi lager et nytt objekt av klassen, og brukes til å definere hvilke variabler objektet skal ha fra starten.

`self` er ordet vi bruker for å vise at en variabel eller metode hører til **dette spesifikke objektet**.

**Navnekonvensjon:** Navnet på en klasse skal alltid starte med stor bokstav, f.eks. `Person`, `Bok`, `Bil`.

```python
class Person:
    def __init__(self, navn, alder):   # __init__() kjøres automatisk når vi lager et objekt
        self.navn = navn
        self.alder = alder

# Lager to objekter (instanser) av klassen Person
p1 = Person("Markus", 25)
p2 = Person("Aksel", 16)

print(p1.navn)
print(p2.navn)
print(p1.alder)
```

Legg merke til at `p1` og `p2` er **to helt separate objekter**, selv om de er laget av samme klasse. De har hver sine egne verdier for `navn` og `alder`.

## 10.3 Metoder - funksjoner inni en klasse

En klasse kan også inneholde egne funksjoner, kalt **metoder**. En metode fungerer akkurat som en vanlig funksjon, men den har alltid `self` som første argument, slik at den får tilgang til objektets egne variabler.

```python
import numpy as np

class Sirkel:
    def __init__(self, radius):
        self.radius = radius

    def areal(self):                          # Metode som regner ut arealet
        a = np.pi * self.radius**2
        print("Sirkelen har areal", round(a, 2))

    def omkrets(self):                        # Metode som regner ut omkretsen
        o = 2 * np.pi * self.radius
        print("Sirkelen har omkrets", round(o, 2))

sirkel_1 = Sirkel(10)

sirkel_1.areal()      # Kaller på metoden areal() på objektet sirkel_1
sirkel_1.omkrets()
```

Lag en klasse `Rektangel` med `__init__()`-funksjon som tar imot `bredde` og `hoyde`. Lag metodene `areal()` og `omkrets()` som skriver ut henholdsvis arealet og omkretsen til rektangelet. Test klassen med et selvvalgt rektangel.

```python
# Skriv koden din her

class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w*self.h
    def circum(self):
        return self.w*2 + self.h*2

print(Rectangle(2,5).area())
print(Rectangle(2,10).circum())
```

## 10.4 Et mer avansert eksempel: Ansatt

La oss se på et eksempel som viser hvor kraftige klasser kan være. Vi lager en klasse `Ansatt` som automatisk kan generere brukernavn, et startpassord og en jobb-e-post basert på informasjonen som legges inn.

```python
import random

class Ansatt:
    def __init__(self, fornavn, etternavn, fodselsaar):
        self.fornavn = fornavn
        self.etternavn = etternavn
        self.fodselsaar = fodselsaar

    def brukernavn(self):
        print(f"Brukernavn: {self.fornavn}.{self.etternavn}{self.fodselsaar}")

    def start_passord(self):
        tall_1 = random.randint(0, 9)
        tall_2 = random.randint(0, 9)
        tall_3 = random.randint(0, 9)
        print(f"Startpassord: {self.fornavn[0:3]}{self.etternavn[0:3]}{tall_1}{tall_2}{tall_3}")

    def e_post(self):
        print(f"E-post: {self.fornavn[0:3]}{self.etternavn[0:3]}{self.fodselsaar}@jobb.no")


ansatt_1 = Ansatt("Roger", "Pettersen", 1990)

ansatt_1.brukernavn()
ansatt_1.start_passord()
ansatt_1.e_post()
```

Legg merke til hvor mye funksjonalitet vi har samlet i én klasse! Vi kunne laget hundrevis av `Ansatt`-objekter, og hver av dem ville automatisk fått sitt eget brukernavn, passord og e-post - uten at vi trenger å skrive den koden på nytt for hver ansatt.

Lag en klasse `Spiller` (til et enkelt spill) med `navn`, `poeng` (start på 0) og `level` (start på 1). Lag en metode `legg_til_poeng(antall)` som øker poengsummen, og som automatisk øker `level` med 1 hver gang spilleren når et multiplum av 100 poeng (100, 200, 300, osv). Skriv ut en melding når spilleren går opp i level.

```python
# Oppgave E

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
    
    def put_in(self, amount):
        self.balance += amount

    def take_out(self, amount):
        if(self.balance >= amount):
            self.balance -= amount
        else:
            raise Exception("Not enough funds!")

    def show_balance(self):
        print(f"Your balance: ${self.balance}")

account_a = BankAccount("Me", 2500)
account_a.put_in(500)
account_a.take_out(100)
account_a.show_balance()

account_b = BankAccount("Me", 2500)
account_b.take_out(2600)
account_b.show_balance()
```

```python
# Oppgave F


# Lag en klasse `Spiller` (til et enkelt spill) med `navn`, `poeng` (start på 0) og `level` (start på 1).
# Lag en metode `legg_til_poeng(antall)` som øker poengsummen,
# og som automatisk øker `level` med 1 hver gang spilleren når et multiplum av 100 poeng (100, 200, 300, osv).
# Skriv ut en melding når spilleren går opp i level.

class Player:
    def __init__(self, username):
        self.username = username
        self.points = 0
        self.level = 0
    def add_points(self, points):
        self.points += points
        # If we have more than 100 points
        # Increment level and decrease points
        if self.points >= 100:
            self.points -= 100
            self.level += 1
            print(f"Player {self.username} has levelled up to level {self.level}!")
player1 = Player("pissplayer911")

player1.add_points(99)
player1.add_points(2)
player1.add_points(99)

```
