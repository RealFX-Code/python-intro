## 2.1 Feilmeldinger er hjelp, ikke kjeft

Det første du må vite: **alle som programmerer får feilmeldinger hele tiden.** Ikke bare nybegynnere — også folk som har jobbet med dette i tjue år. Det er en helt normal del av å skrive kode, ikke et tegn på at du er dårlig til det.

Det andre du må vite er enda viktigere: **feilmeldingen er Python som prøver å hjelpe deg.** Den forteller deg to ting:

1. **Hva** som er galt
2. **Hvor** i koden det er galt

Klarer du å lese ut de to tingene, har du løst det meste selv. Mange nybegynnere ser en rød feilmelding, blir stresset, og ber om hjelp uten å lese hva det står. Det er synd — svaret står som regel rett der.

## 2.2 Slik leser du en feilmelding

Kjør cella under. Den kommer til å gi en feilmelding — det er meningen.

````python
# ⚠️ Denne cellen skal gi feilmelding. Kjør den og les hva som står.
alder = "16"
neste_ar = alder + 1
print(neste_ar)
````

Du fikk noe som ser omtrent slik ut:

```
Traceback (most recent call last):
  Cell In[1], line 3
    neste_ar = alder + 1
               ~~~~~~^~~
TypeError: can only concatenate str (not "int") to str
```

Det ser skummelt ut, men det er egentlig bare tre nyttige opplysninger. **Les nedenfra og opp:**

### 🔻 Nederste linje: HVA som er galt

```
TypeError: can only concatenate str (not "int") to str
```

Dette er den viktigste linja. Den består av to deler:

- **`TypeError`** — feiltypen. Her: noe har feil datatype.
- **Forklaringen etter kolonet** — her: du kan bare sette sammen tekst med tekst, ikke tekst med et heltall.

I dette tilfellet: `alder` er teksten `"16"`, ikke tallet `16`. Python nekter å legge sammen tekst og tall.

### 🔺 Linjene over: HVOR feilen er

```
  Cell In[1], line 3
    neste_ar = alder + 1
               ~~~~~~^~~
```

- **`line 3`** — feilen er på linje 3 i cella
- Under står selve kodelinja, så du slipper å telle
- **`^`-tegnet** peker på nøyaktig hvor i linja det knakk

### Oppskriften

1. Se på **nederste linje** → hva er galt?
2. Se på **linjenummeret** → hvor er det galt?
3. Gå til den linja og se etter.

## 2.3 De vanligste feiltypene

Nå går vi gjennom feiltypene du kommer til å møte oftest. For hver av dem: hva den betyr, en celle som utløser den, og hvordan du fikser det.

### SyntaxError — Python skjønner ikke hva du har skrevet

Dette er skrivefeil i selve språket: en parentes som ikke er lukket, eller et anførselstegn som mangler.

**Spesielt for SyntaxError:** ingenting i cella kjøres. Python leser koden før den kjører den, og stopper med en gang den ser noe den ikke forstår.

````python
# ⚠️ Denne cellen skal gi feilmelding.
print("Hei verden"
````

**Hva var galt?** Parentesen ble aldri lukket. Feilmeldingen sier det rett ut:

```
SyntaxError: '(' was never closed
```

Prøv nå den neste — her er det anførselstegnet som mangler.

````python
# ⚠️ Denne cellen skal gi feilmelding.
print("Hei verden)
````

```
SyntaxError: unterminated string literal
```

*Unterminated string* betyr «tekst som aldri ble avsluttet». Python leter etter det andre anførselstegnet og finner det aldri.

De vanligste årsakene til SyntaxError akkurat nå:
- Parentes som ikke er lukket: `print("hei"`
- Anførselstegn som mangler: `print("hei)`
- Manglende komma mellom argumenter: `print("Jeg er" 16 "år")`

> 💡 Står feilen på en linje som ser helt riktig ut? Se på **linja over**. Glemmer du en parentes på linje 4, oppdager ofte ikke Python det før linje 5.

Senere i kurset lærer du om kolon (`:`) og innrykk. Da får du en ny variant av SyntaxError når kolonet mangler — men prinsippet for å lese feilmeldingen er akkurat det samme.

### IndentationError — feil innrykk

Python bryr seg om mellomrommene i starten av en linje. Er det mellomrom der det ikke skal være, klager Python.

Akkurat nå skal ingen av linjene dine ha innrykk i det hele tatt. Senere, når du lærer løkker og funksjoner, skal noen linjer rykkes inn med vilje.

````python
# ⚠️ Denne cellen skal gi feilmelding.
tall = 5
   print(tall)
````

**Hva var galt?** Det er tre mellomrom foran `print(tall)`. Feilmeldingen sier:

```
IndentationError: unexpected indent
```

*Unexpected indent* = «uventet innrykk». Python skjønner ikke hvorfor den linja er dyttet inn.

Dette skjer oftest ved et uhell, for eksempel når du limer inn kode fra et nettsted og får med noen mellomrom på kjøpet.

### NameError — Python kjenner ikke navnet

Du bruker et variabelnavn som ikke finnes. Nesten alltid en skrivefeil, eller at du har glemt å kjøre cella der variabelen ble laget.

**Husk at Python skiller mellom store og små bokstaver:** `navn` og `Navn` er to helt forskjellige variabler.

````python
# ⚠️ Denne cellen skal gi feilmelding.
navn = "Kari"
print("Hei", Navn)
````

**Hva var galt?** Variabelen heter `navn` med liten n, men vi skrev `Navn` med stor N.

Legg merke til at Python ofte gjetter for deg:

```
NameError: name 'Navn' is not defined. Did you mean: 'navn'?
```

Den siste delen — *Did you mean* — er Python som foreslår hva du sannsynligvis mente. Det er verdt å lese.

> 💡 **Vanlig felle i notebooks:** får du NameError på en variabel du *vet* du har laget, har du sannsynligvis ikke kjørt cella der den ble laget. Cellene må kjøres i rekkefølge. Velg **Run → Run All Above** for å kjøre alt over.

### TypeError — feil datatype

Du prøver å gjøre noe med en datatype som ikke støtter det. Klassikeren er å blande tekst og tall.

````python
# ⚠️ Denne cellen skal gi feilmelding.
antall_varer = "5"
pris_per_vare = 30
total = antall_varer * pris_per_vare
print(total + 100)
````

**Hva var galt?** `antall_varer` er teksten `"5"`, ikke tallet 5.

Her er noe interessant: linje 3 krasjet **ikke**. Python gjentok bare teksten `"5"` tretti ganger, som er noe helt annet enn å regne 5 · 30. Feilen dukket først opp på linje 4, da vi prøvde å legge 100 til en tekst.

Det er en nyttig påminnelse: **feilen oppstår ikke alltid der årsaken ligger.** Peker Python på linje 4, kan årsaken godt være noe du gjorde på linje 1.

Riktig kode ville vært:

```python
antall_varer = 5        # tall, ikke tekst
```

> 💡 Er du usikker på hvilken datatype noe har? Bruk `type()`: `print(type(antall_varer))`

### ValueError — riktig datatype, men umulig verdi

Her er datatypen riktig, men selve verdien gir ikke mening for det du prøver å gjøre.

````python
# ⚠️ Denne cellen skal gi feilmelding.
tall = int("tolv")
print(tall)
````

**Hva var galt?** `int()` kan gjøre om `"42"` til tallet 42, men `"tolv"` er ikke et tall skrevet med siffer.

```
ValueError: invalid literal for int() with base 10: 'tolv'
```

*Invalid literal* betyr «ugyldig verdi». Python sier: jeg forstår at du vil ha et heltall, men dette klarer jeg ikke å tolke som et tall.

**Forskjellen på TypeError og ValueError:**
- **TypeError** — feil *type*: `"5" + 3`
- **ValueError** — riktig type, umulig *verdi*: `int("tolv")`

### ZeroDivisionError — deling på null

Denne er akkurat det den ser ut som. Matematikken tillater ikke å dele på null, og det gjør ikke Python heller.

````python
# ⚠️ Denne cellen skal gi feilmelding.
sum_poeng = 240
antall_elever = 0
gjennomsnitt = sum_poeng / antall_elever
print(gjennomsnitt)
````

**Hva var galt?** Vi delte på `antall_elever`, som var 0.

Dette virker kanskje som en dum feil å gjøre med vilje, men i ekte programmer skjer det hele tiden: koden virker fint helt til telleren tilfeldigvis står på null. Da må du sjekke verdien før du deler — det lærer du hvordan i notebook 3.
