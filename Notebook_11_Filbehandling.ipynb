## 11.1 Hvorfor filbehandling?

Så langt har alt vi har jobbet med "forsvunnet" når vi lukker programmet - variabler og lister lagres bare mens programmet kjører. Skal vi **lagre data permanent**, eller **lese inn data** noen andre har samlet inn (for eksempel værdata, høyscore-lister eller kundeinformasjon), må vi lese og skrive til **filer**.

### Modus (mode) ved åpning av fil

Når vi åpner en fil med `open()`, må vi si hva vi skal gjøre med den:

| Modus | Betydning |
|---|---|
| `"r"` | Read - les fra filen (standardvalg) |
| `"w"` | Write - skriv til filen. **Overskriver** alt som står der fra før! |
| `"a"` | Append - legger til tekst **bakerst** i filen, uten å slette det som var der |
| `"rt"` | Read text - les filen som vanlig tekst |

## 11.2 Skrive til en fil

Vi åpner en fil for skriving med `open("filnavn.txt", "w")`, skriver med `.write()`, og lukker filen med `.close()` når vi er ferdige. Dersom filen ikke finnes fra før, blir den opprettet automatisk.

```python
f = open("dagbok.txt", "w")
f.write("Dette er min første linje i filen.")
f.close()

print("Filen er skrevet!")
```

### 11.2.1 `with open(...)` - en tryggere måte å jobbe med filer på

Bruker vi `with open(...) as file:`, trenger vi **ikke** å huske å skrive `.close()` selv - Python lukker filen automatisk når vi er ferdige med koden inni `with`-blokken. Dette er den vanligste og tryggeste måten å jobbe med filer på.

```python
with open("dagbok.txt", "w") as file:
    file.write("Dette er min første linje i filen.")

print("Filen er skrevet, og lukket automatisk!")
```

## 11.3 Lese fra en fil

For å lese innholdet i en fil, åpner vi den i lesemodus (`"r"` eller `"rt"`).

| Metode | Effekt |
|---|---|
| `.read()` | Leser **hele** filen som én lang tekststreng |
| `.readline()` | Leser **én** linje om gangen |
| `.readlines()` | Leser alle linjene, og gir tilbake en **liste** der hver linje er ett element |

```python
with open("dagbok.txt", "r") as file:
    innhold = file.read()

print(innhold)
```

## 11.4 Flere linjer, og å legge til tekst (append)

Vi bruker `\n` for å lage et linjeskift inni teksten vi skriver. Bruker vi modus `"a"` (append) i stedet for `"w"`, legges den nye teksten **til bakerst**, i stedet for å overskrive filen.

```python
with open("dagbok.txt", "w") as file:
    file.write("Dag 1: Startet med Python i dag.\n")
    file.write("Dag 2: Lærte om lister.\n")

with open("dagbok.txt", "a") as file:      # "a" for append - legger til, sletter ikke
    file.write("Dag 3: Lærte om filbehandling!\n")

with open("dagbok.txt", "r") as file:
    print(file.read())
```

### Lese linje for linje med en for-løkke

En fil kan gås gjennom med en `for`-løkke, akkurat som en liste. Da får vi én linje av gangen, inkludert linjeskiftet (`\n`) på slutten av hver linje. Vi bruker ofte `.strip()` for å fjerne dette linjeskiftet (og eventuelt mellomrom) fra hver linje.

```python
with open("dagbok.txt", "r") as file:
    for linje in file:
        linje = linje.strip()      # Fjerner linjeskift og mellomrom i start/slutt
        print("Logg:", linje)
```

## 11.5 CSV-filer

**CSV** står for "Comma-Separated Values" - en enkel filtype der data er organisert i rader og kolonner, atskilt med komma (`,`) eller semikolon (`;`). CSV-filer er svært vanlige for å lagre og dele tabelldata, for eksempel fra Excel.

La oss lage en enkel CSV-fil med poengsummene til noen spillere, og deretter lese den inn og tolke innholdet selv.

```python
with open("spillere.csv", "w") as file:
    file.write("navn,poeng\n")
    file.write("Ola,45\n")
    file.write("Kari,78\n")
    file.write("Nils,62\n")
    file.write("Mona,90\n")

with open("spillere.csv", "r") as file:
    print(file.read())
```

For å bruke dataene i Python, må vi selv **splitte** hver linje på komma med `.split(",")`, slik at vi får hver verdi separat.

```python
datasett = []

with open("spillere.csv", "r") as file:
    for linje in file:
        linje = linje.strip()          # Fjerner linjeskift
        rad = linje.split(",")          # Deler linjen opp der det står komma
        datasett.append(rad)

print(datasett)
```

Legg merke til at den første raden (`["navn", "poeng"]`) er **overskriften**, ikke selve dataene. Vi kan hoppe over den når vi jobber videre med tallene.

```python
overskrift = datasett[0]
data = datasett[1:]               # Alle rader unntatt overskriften

print("Overskrift:", overskrift)

for rad in data:
    navn = rad[0]
    poeng = int(rad[1])            # Gjør om teksten til et heltall
    print(navn, "har", poeng, "poeng")
```

### 11.5.1 Python-biblioteker for filbehandling

Å lese filer "for hånd" med `.split()` fungerer fint til enkle filer, men Python har også egne biblioteker som gjør jobben enklere for større datasett:

- **`csv`** - et innebygd bibliotek spesielt laget for å lese og skrive CSV-filer
- **`numpy`** - `np.genfromtxt("fil.csv", delimiter=",")` leser tall direkte inn i et array
- **`pandas`** - `pd.read_csv("fil.csv")` er kraftig verktøy for å jobbe med større datasett, og brukes mye i databehandling og dataanalyse

Disse bibliotekene lærer du mer om etter hvert som datasettene du jobber med blir større.

```python
import csv

with open("spillere.csv", "r") as file:
    leser = csv.reader(file)
    for rad in leser:
        print(rad)
```

## 11.6 Slette filer

Biblioteket `os` gir oss tilgang til funksjoner som jobber med selve operativsystemet, blant annet `os.remove()` for å slette en fil permanent.

**NB:** Vær forsiktig med `os.remove()` - filen blir slettet med det samme, uten noen bekreftelse eller papirkurv å hente den fra!

```python
import os

with open("midlertidig.txt", "w") as file:
    file.write("Dette er en midlertidig fil.")

print("midlertidig.txt" in os.listdir())   # Sjekker om filen finnes i mappen

os.remove("midlertidig.txt")

print("midlertidig.txt" in os.listdir())   # Filen er nå borte
```

### 🔴 Vanskelig nivå (utfordrende)

**Oppgave E**  
Lag en CSV-fil `karakterer.csv` med kolonnene `navn,fag,karakter` for minst 6 rader (noen elever kan gjerne ha flere fag hver). Skriv et program som:
1. Leser filen inn i en liste
2. Regner ut gjennomsnittskarakteren for **hver elev**
3. Skriver ut navnet og gjennomsnittet, sortert fra høyest til lavest snitt

*Tips: en dictionary kan være nyttig her - bruk elevnavnet som nøkkel, og lagre alle karakterene til hver elev i en liste som verdi.*

**Oppgave F**  
Lag et program som simulerer en enkel "høyscore-fil" for et spill:
1. Hvis filen `highscore.txt` ikke finnes fra før, opprett den med teksten `"0"`
2. Les inn nåværende høyscore fra filen
3. Bruk `random.randint()` til å simulere en ny poengsum
4. Dersom den nye poengsummen er høyere enn den lagrede, skriv den nye høyscoren til filen og skriv ut en gratulasjonsmelding
5. Kjør programmet flere ganger, og se om høyscoren endrer seg over tid!

```python
# Oppgave E
from random import randint, choice
import csv

# Spaghetti code!

with open("Karakterer.csv", "w") as f:
    names = [ "Jon", "Albert", "Maja", "Elisa" ]
    fag = [ "Matematikk", "Norsk", "Programmering", "Engelsk", "Teknologiforståelse" ]
    f.write("navn,fag,karakter\n")
    # For each student
    for name in names:
        # Store what subjects we've done
        done_fag = []
        # Choose a random amount of subjects
        for num in range(randint(1, len(fag))):
            while True:
                # Chose a random subject
                randomfag = choice(fag)
                # If we have done that subject, generate a random grade.
                # Don't match for ones we've done, just let it loop
                # until we have one we haven't done yet.
                if randomfag not in done_fag:
                    f.write(f"{name},{randomfag},{randint(1,6)}\n")
                    break

# Example format:
# grades = {
#     "Jon": [
#         1,
#         5,
#         3
#     ]
# }

grades = {}

with open("Karakterer.csv", "r") as f:
    csvreader = csv.reader(f)
    for row in csvreader:
        # Skip header, we know the fields!
        if(row[0] == "navn"):
            continue
        # Student name as key
        # If we don't have an entry with the student name, make one!
        if row[0] not in grades:
            grades[row[0]] = []
        # Discard the subject, we don't need it.
        # grades[row[0]].append({row[1], row[2]})
        grades[row[0]].append(int(row[2]))

averages = []


for name in grades:
    average = round(sum(grades[name]) / len(grades[name]), 2)
    averages.append([ name, average])
    print(f"Gjennomsnittet til {name} er {average}!")

print("\nSortert fra høyest til lavest snitt:")

for entry in sorted(averages, key=lambda x: x[1], reverse=True):
    print(f"Gjennomsnittet til {entry[0]} er {entry[1]}!")
```

```python
# Oppgave F
import random
import os

filename = "highscore.txt"

currenthighscore = 0

if not os.path.isfile(filename):
    # Create empty file if it doesn't exist
    with open(filename, "w") as f:
        f.write(str(currenthighscore))


with open(filename, "r") as f:
    # Read our saved highscore
    currenthighscore = int(f.read())

# New score is between 1 and 2000
newscore = random.randint(1,2000)

print(f"Highscore: {currenthighscore} | Your score: {newscore}")

if newscore > currenthighscore:
    print(f"Congratulations! New High-Score: {newscore}")
    with open(filename, "w") as f:
        f.write(str(newscore))
```
